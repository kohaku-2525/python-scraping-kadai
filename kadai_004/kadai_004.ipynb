**Reasoning**:
Based on the HTML inspection, the stock data is found within a table with class 'table yf-1mr6o9m'. I will now update the `extract_stock_data` function to parse this table, iterate through its rows, extract the date, open, high, low, and close values from the table data cells, format the date, convert prices to floats, and return the data as a list of lists.



```python
from bs4 import BeautifulSoup
from datetime import datetime

def extract_stock_data(html_content):
    """
    Extracts stock data (date, open, high, low, close) from HTML content.
    This version is tailored for the Nikkei 225 page structure found.

    Args:
        html_content (str): The HTML content of the webpage.

    Returns:
        list: A list of lists, where each inner list contains [date, open, high, low, close].
              Date is formatted as 'YYYY-MM-DD'.
    """
    soup = BeautifulSoup(html_content, 'html.parser')
    stock_data = []

    # The previous inspection identified the first table as containing the data.
    # It does not have a specific class, so we target the first table found.
    historical_data_table = soup.find('table')

    if historical_data_table:
        # The data is in the first <tr> within the <tbody>
        # We are looking for data in the first row of the table, not headers.
        rows = historical_data_table.find_all('tr')

        if rows:
            # The first row contains the actual stock data for a single day
            data_row = rows[0]
            columns = data_row.find_all('td')

            # Ensure there are enough columns for Date, Open, High, Low, Close
            if len(columns) >= 5:
                try:
                    date_raw = columns[0].get_text(strip=True)
                    # Example date_raw: '2025/12/5'
                    dt_object = datetime.strptime(date_raw, '%Y/%m/%d')
                    formatted_date = dt_object.strftime('%Y-%m-%d')

                    # Prices are prefixed with Japanese labels, e.g., '始値: 50,530.34'
                    open_price_str = columns[1].get_text(strip=True).split(':')[1].replace(',', '')
                    high_price_str = columns[2].get_text(strip=True).split(':')[1].replace(',', '')
                    low_price_str = columns[3].get_text(strip=True).split(':')[1].replace(',', '')
                    close_price_str = columns[4].get_text(strip=True).split(':')[1].replace(',', '')

                    # Convert prices to float, handling potential non-numeric values
                    open_price = float(open_price_str) if open_price_str != '-' else None
                    high_price = float(high_price_str) if high_price_str != '-' else None
                    low_price = float(low_price_str) if low_price_str != '-' else None
                    close_price = float(close_price_str) if close_price_str != '-' else None

                    stock_data.append([formatted_date, open_price, high_price, low_price, close_price])
                except (ValueError, IndexError, AttributeError) as e:
                    print(f"Error parsing data row: {data_row.get_text(strip=True)}. Error: {e}")
        else:
            print("No data rows found in the table.")
    else:
        print("Historical data table not found.")

    return stock_data

print("Function `extract_stock_data` updated with Nikkei 225 parsing logic.")
```
