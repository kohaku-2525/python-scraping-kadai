```python
import requests
from bs4 import BeautifulSoup

url = 'https://news.yahoo.co.jp/articles/2efdde68b40d425c5fd765ccdc41ac8db3804d22'

response = requests.get(url)
soup = BeautifulSoup(response.text, 'html.parser')

article_body_elements = soup.select('div.article_body p')

article_text = []
for p_tag in article_body_elements:
    article_text.append(p_tag.get_text())

full_article = '\n'.join(article_text)
print(full_article)
```
